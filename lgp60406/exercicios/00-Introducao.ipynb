## Olá Mundo!

- Tradicionalmente o primeiro programa escrito é um __``Olá Mundo''__.
- Para introduzir você ao mundo em C:

```c++
/* Programa básico que exibe "olá mundo" */

/* Bibliotecas */
#include <stdio.h>


/* Função main: ponto de entrada */
int main() {

	// Imprime Olá mundo!!!
	printf("Olá mundo!!!\n");

	return 0;
}
```

## Comentários

- Comentários são ignorados pela compilador quando um programa é executado.
- Eles ajudam na compreensão do código.
- Comentários em C são ativados por:
    - linha: //
    - Bloco: /*     */ 

```c++
/* Exemplos de comentários */

/* Bibliotecas */
#include <stdio.h>

/* Função main: ponto de entrada */
int main() {

	// Isso é um comentário
    printf("Olá!\n");  // A partir daqui também é!

    printf("isso não é comentário.\n");
    /* Um novo comentário pode ser colocado aqui!
     * Em bloco, ou seja, mais de uma linha       */

    return 0;
}
```

# A função _printf_

- A função __printf__ imprime dados do programa.
- Ela está definida na biblioteca __stdio.h__
    - printf: _character out_
- Os dados colocados entre parênteses são conhecidos como __argumentos__ e são enviados ao console..
    - O console é a __saída padrão__ do programa.

```c++
/* Exemplo com impressão de caracteres especiais */

/* Bibliotecas */
#include <stdio.h>

/* Função main: ponto de entrada */
int main() {

    // Se eu precisar imprimir o caractere "
    // Devo colocar uma barra na frente
    printf("Desejo imprimir o \" por algum motivo.\n");

    // E se eu quiser o caractere \ ?
    printf("Esse arquivo está em C:\\minha_pasta\n");

    return 0;
}
```

| Caractere | Descrição     |
|  :---:    |   :---:       |
| \\'       | Aspas simples |
| \\"       | Aspas duplas  |
| \t        | Tabulação     |
| \n        | Nova linha    |

# Variáveis e entrada de dados

- Uma __variável__ é utilizada para guardar dados que serão reutilizados.
- As variáveis em C devem ser criadas de forma explícita contendo seu __tipo__ e __nome__

```c++
/* Bibliotecas */
#include <stdio.h>

/* Função main: ponto de entrada */
int main() {
    int numero_inteiro = 5;          //inteiro, int ou integer
    float numero_virgula  =  9.69;   //real ou float
    char caractere = 'O';         //caractere, sempre entre aspas simples

    return 0;
}
```

- Existem três tipos básicos de variáveis em C:
    - __Caracter__: dado vindo do teclado, por exemplo. 
    - __Números inteiros__: números sem ponto. Conhecidos como __int__ ou __integer__.
    - __Números reais__: números com ponto. Conhecidos como __float__.

- Tanto o printf quanto o scanf utilizam o marcador __%__ para converter os dados vindos do teclado (texto) para formato binário.

| Marcado  | Descrição                                  |
|  :---:    |   :---:                                   |
| ``%d``    | Converte/imprime como número inteiro     |
| ``%f``    | Converte/imprime como número real (float) |
| ``%c``    | Mantém como caractere (sem conversão)     |

```c++
/* Entrada de dados e tipos de variáveis */

/* Bibliotecas */
#include <stdio.h>

/* Função main: ponto de entrada */
int main() {
	int x;
	float y;
	char z;

	printf("Entre com um (inteiro real caractere): ");
	/* A entrada de dados é separada por espaço, exemplo: 43 56.7 j
    scanf("%d %f %c", &x, &y, &z);

	printf("\n");

	printf("Inteiro: %d\n"
		   "Real: %f\n"
		   "caracter: %c\n",
		   x,y,z);

	return 0;
}
```

# Operadores
Se os dados são numéricos, você pode realizar as seguintes operações:


| Operador  | Descrição                          |
|  :---:    |   :---:       |
| ``+``     | Adição                             |
| ``-``     | Subtração                          |
| ``*``     | Multiplicação                      |
| ``/``     | Divisão                            |
| ``%``     | Módulo (Resto de uma divisão inteira) |

Exemplos:

```c++
/* Bibliotecas */
#include <iostream>

/* Namespace padrão */
using namespace std;

/* Função main: ponto de entrada */
int main() {
	int x,y,z;
    float a;
	int b,c;

	x = 3 + 4;   // x vale = 7

	x = 5;
	y = 6;
	z = x - 2 * y;   // z = 5 - 6 * 2 = -7 (cuidado com a
	                 // precedência dos operadores)
	x = 8;
	y = 6;
	a = (float)x / y;  // a = 1.3333333333333333
	b = x / y;         // b = 1
	c = x % y;         //c = 2
}
```

| Operador  | Descrição    |
|  :---:    |   :---:      |
| ``>``     | Maior        |
| ``<``     | Menor        |
| ``>=``    | Maior igual  |
| ``<=``    | Menor igual  |
| ``==``    | Igual        |
| ``!=``    | Diferente    |

```c++
/* Programa básico que exibe "olá mundo" */

/* Bibliotecas */
#include <stdio.h>


/* Função main: ponto de entrada */
int main() {
	/* Variáveis */
	int r=-1;
	int s=1;
	int t=2;

	/* Simbolos de conversao
	Simbolo
	    !         negação
	   &&         and
	   ||         or */

	/* Operadores
	 >    maior
	 <    menor
	 <=   menor igual
	 >=   maior igual
	 ==   igual
	 !=   diferente */

	/* Letra a */
	bool a = (r+5)>t;
	printf("%d\n", a);

	/* Letra b */
	// (...)

	/* Letra c */
	bool c = (-r == s) && (s > r);
	printf("%d\n", c);

	return 0;
}
```

| Operador  | Descrição     |
|  :---:    |   :---:       |
| ``\|\|``  | OU            |
| ``&&``    | E             |
| ``!``     | NÃO           |
| ``++``    | Incremento    |
| ``--``    | Decremento    |
